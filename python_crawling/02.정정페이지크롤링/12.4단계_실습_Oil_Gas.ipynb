```python
#20250707_지선 부장님 요청_정적 크롤링
import requests
from bs4 import BeautifulSoup
import pandas as pd

data = []

for i in range(1, 331):  # 1~5페이지 크롤링 (원하는 만큼 조절)
    url = f"https://www.shalexp.com/companies?page={i}"
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')
    rows = soup.select("tbody tr")  # 표 행들 선택

    for row in rows:
        tds = row.find_all("td")
        if len(tds) < 5:
            continue  # 빈칸 포함 5개 td가 있어야 정상 데이터

        a_tag = tds[1].find("a")
        if not a_tag:
            continue  # <a> 태그 없으면 건너뜀

        company_name = a_tag.text.strip()
        company_link = a_tag['href'].strip()
        if company_link.startswith('/'):
            company_link = "https://www.shalexp.com" + company_link

        company_location = tds[2].text.strip()
        company_type = tds[3].text.strip()
        most_recent_production = tds[4].text.strip()

        print(company_name, company_link, company_location, company_type, most_recent_production)
        data.append([company_name, company_link, company_location, company_type, most_recent_production])

# DataFrame 생성
df = pd.DataFrame(data, columns=[
    "Company Name", "Website", "Company Location", "Company Type", "Most Recent Production"
])

# 결과 출력
print(df)

# CSV 저장 (원하는 경우)
df.to_csv("shalexp_companies2.csv", index=False)
```
