```python
import whisper

# 모델 로드 (small/medium/large 선택 가능, small 추천)
model = whisper.load_model("small")

# 동영상에서 영어 텍스트 추출 (언어는 'en'으로 고정)
result = model.transcribe(r"C:\Temp\C_movie\20250610_110915.mp4", language="en")#파일 경로와 언어만 변경

# 전체 텍스트 출력
print(result['text'])

# 파일로 저장 (옵션)
with open("output.txt", "w", encoding="utf-8") as f:
    f.write(result['text'])
```
