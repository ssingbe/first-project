```python
#엑셀 파일 합치기(초기 버전으로 느림)
import os
import pandas as pd

# 엑셀 파일이 있는 폴더 경로
folder_path = r'C:\Users\DTE24089\Desktop\250522이후 파일\2503_1'  # ← 경로 수정

# 결과 저장 파일 이름
output_file = 'merged_with_filename2.xlsx'

# 엑셀 파일 목록 필터링
excel_files = [f for f in os.listdir(folder_path) if f.endswith('.xlsx') and not f.startswith('~')]

# 데이터 병합용 DataFrame 초기화
merged_df = pd.DataFrame()

for file in excel_files:
    file_path = os.path.join(folder_path, file)
    try:
        # 엑셀 파일 읽기
        df = pd.read_excel(file_path, engine='openpyxl')
        # 파일명에서 확장자 제거 후 열 추가
        df['source_file'] = os.path.splitext(file)[0]
        # 병합
        merged_df = pd.concat([merged_df, df], ignore_index=True)
        print(f'✔ 처리 완료: {file}')
    except Exception as e:
        print(f'⚠ 오류 발생: {file} - {e}')

# 결과 저장
output_path = os.path.join(folder_path, output_file)
merged_df.to_excel(output_path, index=False)
print(f'\n✅ 병합 완료 → {output_path}')
```

```python
#엑셀 파일 합치기(순차 버전으로 느림)
import os
import pandas as pd
import time

# 설정
folder_path = r'C:\Users\DTE24089\Desktop\250522이후 파일\2503'
output_file = os.path.join(folder_path, "merged_result_소압2.xlsx")

failed_files = []
df_list = []

start_time = time.time()
print("🚀 Excel 병합 시작 (순차 버전)")

# 대상 파일 목록
xlsx_files = [os.path.join(folder_path, f)
              for f in os.listdir(folder_path)
              if f.endswith('.xlsx') and not f.startswith('~')]

for file_path in xlsx_files:
    base = os.path.basename(file_path)
    try:
        print(f"🔄 병합 중: {base}")
        df = pd.read_excel(file_path, engine='openpyxl')
        df['source_file'] = os.path.splitext(base)[0]
        df_list.append(df)
    except Exception as e:
        print(f"❌ 실패: {base} - {e}")
        failed_files.append(base)

# 병합 및 저장
if df_list:
    merged_df = pd.concat(df_list, ignore_index=True)
    merged_df.to_excel(output_file, index=False)
    print(f"\n✅ 병합 완료: {output_file}")
    print(f"📊 총 행 수: {len(merged_df):,}, 열 수: {len(merged_df.columns)}")
else:
    print("❗ 병합할 데이터가 없습니다.")

# 실패 파일 출력
if failed_files:
    print(f"\n⚠ 실패한 파일 목록 ({len(failed_files)}개):")
    for f in failed_files:
        print(f" - {f}")
else:
    print("\n🎉 모든 파일 정상 처리 완료!")

print(f"\n⏱ 총 소요 시간: {round(time.time() - start_time, 2)}초")
```
