```python
# ==============================================================================
# Step 0. [환경설정] - 한글 폰트/필수 패키지 임포트
# ==============================================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
import seaborn as sns
import os
import re
from typing import Dict, List, Optional, Tuple
from math import ceil

# 한글 폰트 설정 (Windows: Malgun Gothic, Mac: AppleGothic)
try:
    matplotlib.rc('font', family='Malgun Gothic')
except:
    matplotlib.rc('font', family='AppleGothic')
matplotlib.rc('axes', unicode_minus=False)

# ==============================================================================
# Step 1. [분석 함수 정의]
# ==============================================================================
def find_header_row(file_path: str, sheet_name: str) -> int:
    """
    엑셀 시트에서 'time' 문자열이 포함된 행을 찾아 헤더로 사용될 행 번호를 반환합니다.
    """
    try:
        tmp_df = pd.read_excel(file_path, sheet_name=sheet_name, engine='openpyxl', header=None, nrows=20)
        for i, row in tmp_df.iterrows():
            if any('time' in str(cell).strip().lower() for cell in row):
                return i
    except Exception as e:
        print(f"'{sheet_name}' 시트에서 헤더를 찾는 중 오류 발생: {e}")
    return 0

def extract_bloom_sensor_table(df: pd.DataFrame, sheet_name: str, time_point: float) -> Optional[pd.DataFrame]:
    """
    주어진 데이터프레임에서 특정 시점(time_point)의 Bloom 센서 데이터를 추출하여 테이블 형태로 가공합니다.
    """
    try:
        time_col = next((col for col in df.columns if 'time' in str(col).lower()), None)
        if not time_col:
            print(f"'{sheet_name}' 시트에서 Time 컬럼을 찾을 수 없습니다.")
            return None

        df[time_col] = pd.to_numeric(df[time_col], errors='coerce')
        if df[time_col].isna().all():
            print(f"'{sheet_name}' 시트의 Time 컬럼에 유효한 숫자 데이터가 없습니다.")
            return None
        
        time_data_row = df.iloc[df[time_col].sub(time_point).abs().idxmin()]

        rows = []
        for i in range(1, 19):
            for j in range(1, 10):
                col_name = f'Bloom {i} sensor {j}'
                if col_name in df.columns:
                    value = time_data_row.get(col_name)
                    if pd.notna(value):
                        rows.append({"Case": sheet_name, "Time": time_point, "Bloom": i, "Sensor": j, "Value": value})

        return pd.DataFrame(rows) if rows else None
    except Exception as e:
        print(f"'{sheet_name}' 시트, {time_point}초 시점 데이터 추출 중 오류: {e}")
        return None

def extract_cooling_curve_data(df: pd.DataFrame, sheet_name: str, start_time: float, duration: float) -> Optional[pd.DataFrame]:
    """
    냉각 시작 시점부터 지정된 시간 동안의 온도 변화 데이터를 추출합니다.
    """
    try:
        time_col = next((col for col in df.columns if 'time' in str(col).lower()), None)
        if not time_col: return None

        df[time_col] = pd.to_numeric(df[time_col], errors='coerce')
        
        cooling_df = df[(df[time_col] >= start_time) & (df[time_col] <= start_time + duration)].copy()
        if cooling_df.empty: return None

        cooling_df['경과 시간(초)'] = cooling_df[time_col] - start_time
        
        sensor_cols = [col for col in df.columns if 'sensor' in str(col).lower()]
        cooling_df['평균 온도'] = cooling_df[sensor_cols].mean(axis=1)
        
        sampled_times = np.arange(0, duration + 1, 20)
        resampled_data = []
        for t in sampled_times:
            closest_row = cooling_df.iloc[cooling_df['경과 시간(초)'].sub(t).abs().idxmin()]
            resampled_data.append(closest_row)
        
        result_df = pd.DataFrame(resampled_data)[['경과 시간(초)', '평균 온도']]
        result_df['Case'] = sheet_name
        return result_df

    except Exception as e:
        print(f"'{sheet_name}'에서 냉각 곡선 데이터 추출 중 오류: {e}")
        return None


def natural_sort_key(s: str) -> tuple:
    """
    "Case 1", "Case 10", "Case 2*"와 같은 문자열을 자연스럽게 정렬하기 위한 키를 생성합니다.
    """
    match = re.match(r'Case (\d+)(.*)', s)
    if match:
        number, suffix = match.groups()
        return int(number), suffix.strip()
    return (float('inf'), s)

# ==============================================================================
# Step 2. [시각화 함수 정의]
# ==============================================================================
def plot_grouped_comparison_heatmaps(summary_df: pd.DataFrame, meta_df: pd.DataFrame, group_by_col: str, time_desc: str, vmin: float, vmax: float):
    """
    지정한 컬럼으로 케이스를 그룹화하여 히트맵을 시각화합니다.
    """
    groups = summary_df[group_by_col].unique()
    for group in groups:
        cases_in_group = summary_df[summary_df[group_by_col] == group]
        drawable_cases = sorted(cases_in_group[cases_in_group['TimeDesc'] == time_desc]['Case'].unique(), key=natural_sort_key)
        if not drawable_cases:
            continue

        n_cases = len(drawable_cases)
        ncols = min(n_cases, 4)
        nrows = ceil(n_cases / ncols)
        
        fig, axes = plt.subplots(nrows, ncols, figsize=(ncols * 6, nrows * 5))
        fig.suptitle(f'[{time_desc}] {group_by_col}: {group} - Bloom별 평균온도 비교', fontsize=20, y=1.02)
        
        axes = np.array(axes).flatten()
        cmap = plt.get_cmap("YlOrRd").copy()
        cmap.set_bad(color='lightgray')

        for i, case in enumerate(drawable_cases):
            ax = axes[i]
            mat = np.full((2, 6), np.nan)
            labels = np.empty((2, 6), dtype=object)
            
            for bloom_num in range(1, 13):
                row, col = (0, (bloom_num - 7)) if bloom_num > 6 else (1, bloom_num - 1)
                subset = cases_in_group[(cases_in_group['Case'] == case) & (cases_in_group['Bloom'] == bloom_num)]
                if not subset.empty and pd.notna(subset['평균'].values[0]):
                    avg_temp = subset['평균'].values[0]
                    mat[row, col] = avg_temp
                    labels[row, col] = f'{avg_temp:.0f}'
                else:
                    labels[row, col] = 'N/A'

            sns.heatmap(mat, ax=ax, cbar=False, fmt='', cmap=cmap, vmin=vmin, vmax=vmax, linewidths=.5)
            
            for r in range(mat.shape[0]):
                for c in range(mat.shape[1]):
                    if pd.notna(mat[r, c]):
                        normalized_val = (mat[r, c] - vmin) / (vmax - vmin) if (vmax - vmin) > 0 else 0.5
                        font_color = 'white' if normalized_val > 0.6 else 'black'
                        ax.text(c + 0.5, r + 0.5, labels[r, c], ha='center', va='center', fontsize=12, weight='bold', color=font_color)
                    else:
                        ax.text(c + 0.5, r + 0.5, 'N/A', ha='center', va='center', fontsize=12)

            case_info = meta_df[meta_df['Case'] == case].iloc[0]
            title = f"{case}\n({case_info['작업방법']}, {case_info['제어']})"
            ax.set_title(title, fontsize=12)
            ax.set_yticklabels(["상부(7~12)", "하부(1~6)"], rotation=0)
            ax.set_xticks(np.arange(6) + 0.5)
            ax.set_xticklabels([f"위치{i}" for i in range(1, 7)])

        for j in range(n_cases, len(axes)):
            axes[j].set_visible(False)
            
        fig.tight_layout(rect=[0, 0, 0.9, 1])
        sm = plt.cm.ScalarMappable(cmap=cmap, norm=plt.Normalize(vmin=vmin, vmax=vmax))
        sm.set_array([])
        cbar_ax = fig.add_axes([0.92, 0.15, 0.02, 0.7])
        cbar = fig.colorbar(sm, cax=cbar_ax)
        cbar.set_label('평균온도(℃)', size=12)

        plt.show()

def plot_advanced_boxplots(summary_df: pd.DataFrame, time_desc: str, avg_range: tuple, delta_range: tuple, case_order: List[str]):
    """
    서포트 높이 및 제어 방식에 따른 상세 비교 Boxplot을 시각화합니다.
    """
    df_filtered = summary_df[summary_df['TimeDesc'] == time_desc]
    if df_filtered.empty: return

    df_2stack = df_filtered[df_filtered['적치단수'] == '2단']
    df_bottom = df_2stack[df_2stack['Bloom'] <= 6]

    if not df_bottom.empty:
        fig, axes = plt.subplots(1, 2, figsize=(22, 8))
        fig.suptitle(f'[{time_desc}] 서포트 높이별 하부(1단) Bloom 가열 특성 비교', fontsize=18, y=1.02)

        sns.boxplot(ax=axes[0], x='서포트높이', y='평균', data=df_bottom, hue='제어', palette='coolwarm', order=['250mm', '470mm'])
        axes[0].set_title('하부 Bloom 평균 온도 분포', fontsize=14); axes[0].set_ylabel('평균 온도 (°C)'); axes[0].set_ylim(avg_range); axes[0].grid(True, alpha=0.6)

        sns.boxplot(ax=axes[1], x='서포트높이', y='내부ΔT', data=df_bottom, hue='제어', palette='coolwarm', order=['250mm', '470mm'])
        axes[1].set_title('하부 Bloom 내부 온도편차(ΔT) 분포', fontsize=14); axes[1].set_ylabel('내부 온도편차 (°C)'); axes[1].set_ylim(delta_range); axes[1].grid(True, alpha=0.6)

        plt.tight_layout(rect=[0, 0, 1, 0.96]); plt.show()

    if not df_2stack.empty:
        plt.figure(figsize=(18, 9))
        present_cases = df_2stack['Case'].unique()
        plot_order = [case for case in case_order if case in present_cases]
        sns.boxplot(x='Case', y='평균', data=df_2stack, hue='제어', palette='viridis', order=plot_order, dodge=False)
        plt.title(f'[{time_desc}] 제어 방식별 Case별 평균 온도 분포', fontsize=16); plt.ylabel('평균 온도 (°C)'); plt.xticks(rotation=45, ha='right'); plt.ylim(avg_range); plt.grid(True, alpha=0.6); plt.tight_layout(); plt.show()


def plot_case_deviation_boxplots(sensor_df: pd.DataFrame, time_desc: str, target_temp: float, avg_range: tuple, case_order: List[str]):
    """
    (수정) 케이스별 온도 분포, 목표 대비 편차, 내부 Bloom 편차, 전체 편차를 시각화합니다.
    """
    df_filtered = sensor_df[sensor_df['TimeDesc'] == time_desc].copy()
    if df_filtered.empty: return

    df_filtered['목표온도 대비 편차(°C)'] = df_filtered['Value'] - target_temp

    # 내부 Bloom 온도 편차 계산
    bloom_delta_t = df_filtered.groupby(['Case', 'Bloom'])['Value'].agg(lambda x: x.max() - x.min()).reset_index()
    bloom_delta_t.rename(columns={'Value': '내부 Bloom 온도편차(ΔT)'}, inplace=True)

    # 케이스별 전체 온도 편차 (Max-Min) 계산
    case_total_delta_t = df_filtered.groupby('Case')['Value'].agg(lambda x: x.max() - x.min()).reset_index()
    case_total_delta_t.rename(columns={'Value': '전체 온도 편차(Max-Min)'}, inplace=True)

    fig, axes = plt.subplots(2, 2, figsize=(22, 18))
    fig.suptitle(f'[{time_desc}] 케이스별 온도 분포 및 편차 분석', fontsize=18, y=1.0)

    present_cases = df_filtered['Case'].unique()
    plot_order = [case for case in case_order if case in present_cases]

    # 1. 케이스별 전체 센서 온도 분포
    sns.boxplot(ax=axes[0, 0], x='Case', y='Value', data=df_filtered, palette='muted', order=plot_order)
    axes[0, 0].set_title('케이스별 전체 센서 온도 분포', fontsize=14); axes[0, 0].set_ylabel('온도 (°C)'); axes[0, 0].set_xlabel(''); axes[0, 0].set_ylim(avg_range); axes[0, 0].grid(True, alpha=0.6); axes[0, 0].tick_params(axis='x', rotation=90)

    # 2. 케이스별 목표온도 대비 편차 분포
    sns.boxplot(ax=axes[0, 1], x='Case', y='목표온도 대비 편차(°C)', data=df_filtered, palette='muted', order=plot_order)
    axes[0, 1].axhline(0, color='r', linestyle='--', linewidth=2); axes[0, 1].set_title('케이스별 목표온도 대비 편차 분포', fontsize=14); axes[0, 1].set_ylabel('편차 (°C)'); axes[0, 1].set_xlabel(''); axes[0, 1].grid(True, alpha=0.6); axes[0, 1].tick_params(axis='x', rotation=90)

    # 3. 케이스별 내부 Bloom 온도 편차 분포
    sns.boxplot(ax=axes[1, 0], x='Case', y='내부 Bloom 온도편차(ΔT)', data=bloom_delta_t, palette='muted', order=plot_order)
    axes[1, 0].set_title('케이스별 내부 Bloom 온도편차(ΔT) 분포', fontsize=14); axes[1, 0].set_ylabel('편차 (°C)'); axes[1, 0].set_xlabel('Case'); axes[1, 0].grid(True, alpha=0.6); axes[1, 0].tick_params(axis='x', rotation=90)

    # 4. 케이스별 전체 온도 편차 (Max-Min)
    sns.barplot(ax=axes[1, 1], x='Case', y='전체 온도 편차(Max-Min)', data=case_total_delta_t, palette='muted', order=plot_order)
    axes[1, 1].set_title('케이스별 전체 온도 편차 (Max-Min)', fontsize=14); axes[1, 1].set_ylabel('편차 (°C)'); axes[1, 1].set_xlabel('Case'); axes[1, 1].grid(True, alpha=0.6); axes[1, 1].tick_params(axis='x', rotation=90)


    plt.tight_layout(rect=[0, 0, 1, 0.96]); plt.show()

def plot_cooling_curves(cooling_curve_df: pd.DataFrame):
    """
    (수정) 냉각 케이스별 시간에 따른 온도 변화를 1x4 (통합 1개, 개별 3개) 그래프로 시각화합니다.
    """
    if cooling_curve_df.empty:
        print("냉각 곡선 데이터가 없어 그래프를 생성할 수 없습니다.")
        return
    
    cases = sorted(cooling_curve_df['Case'].unique(), key=natural_sort_key)
    
    # 1x4 형태로 subplot 생성
    fig, axes = plt.subplots(1, 4, figsize=(24, 6), sharey=True)
    fig.suptitle('냉각 케이스별 시간에 따른 평균 온도 변화', fontsize=20, y=1.02)
    
    # 1. 통합 그래프 (첫 번째 subplot)
    ax_combined = axes[0]
    sns.lineplot(ax=ax_combined, data=cooling_curve_df, x='경과 시간(초)', y='평균 온도', hue='Case', marker='o', palette='coolwarm')
    ax_combined.set_title('통합 비교', fontsize=16)
    ax_combined.set_xlabel('경과 시간 (초)', fontsize=12)
    ax_combined.set_ylabel('평균 온도 (°C)', fontsize=12)
    ax_combined.grid(True, linestyle='--', alpha=0.7)
    ax_combined.legend(title='Case')

    # 2. 개별 그래프 (나머지 subplots)
    for i, case in enumerate(cases):
        # axes 배열 범위를 벗어나지 않도록 확인
        if i + 1 < len(axes):
            ax_individual = axes[i + 1]
            case_df = cooling_curve_df[cooling_curve_df['Case'] == case]
            
            sns.lineplot(ax=ax_individual, data=case_df, x='경과 시간(초)', y='평균 온도', marker='o', color=f'C{i}')
            
            ax_individual.set_title(f'{case}', fontsize=16)
            ax_individual.set_xlabel('경과 시간 (초)', fontsize=12)
            ax_individual.grid(True, linestyle='--', alpha=0.7)

    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()

# ==============================================================================
# Step 3. [온도 균일도 분석 함수]
# ==============================================================================
def analyze_temperature_uniformity(heating_sensor_df: pd.DataFrame, heating_summary_df: pd.DataFrame, target_temp: float):
    """
    가열 완료 시점의 데이터로 Boxplot 상세 통계치를 포함한 온도 균일도를 분석하고 최적 케이스를 선정합니다.
    """
    
    print("\n" + "="*80 + f"\n[온도 균일도 분석] 목표온도: {target_temp}°C (가열완료 시점 기준)\n" + "="*80)

    sensor_df = heating_sensor_df[heating_sensor_df['TimeDesc'] == '가열완료 시점']
    summary_df = heating_summary_df[heating_summary_df['TimeDesc'] == '가열완료 시점']
    if sensor_df.empty:
        print("온도 균일도 분석을 위한 '가열완료 시점' 데이터가 없습니다.")
        return

    stats_df = sensor_df.groupby('Case')['Value'].describe().reset_index()
    stats_df.rename(columns={
        'mean': '평균', 'std': '표준편차', 'min': '최소값', 
        '25%': 'Q1 (25%)', '50%': '중앙값 (Median)', '75%': 'Q3 (75%)', 'max': '최대값'
    }, inplace=True)
    
    stats_df['최대-최소 차이'] = stats_df['최대값'] - stats_df['최소값']
    stats_df['목표온도와의 차이'] = (stats_df['평균'] - target_temp).abs()

    case_avg_delta_t = summary_df.groupby('Case')['내부ΔT'].mean().reset_index()
    case_avg_delta_t.rename(columns={'내부ΔT': '평균 내부ΔT'}, inplace=True)
    
    analysis_df = pd.merge(stats_df, case_avg_delta_t, on='Case', how='left')

    display_cols = [
        'Case', '평균', '중앙값 (Median)', '표준편차', '최대값', '최소값',
        '최대-최소 차이', 'Q1 (25%)', 'Q3 (75%)', '평균 내부ΔT', '목표온도와의 차이'
    ]
    analysis_df = analysis_df[[col for col in display_cols if col in analysis_df.columns]]

    analysis_df_sorted = analysis_df.sort_values(by='표준편차', ascending=True)

    print("\n[분석 결과 상세 테이블]\n")
    print(analysis_df_sorted.to_string(index=False, float_format='{:.2f}'.format))

    best_case = analysis_df_sorted.iloc[0]
    print("\n\n[최종 결론]")
    print(f"온도 편차가 가장 적은 최적의 케이스는 '{best_case['Case']}' 입니다.")
    print(f"  - 전체 온도 표준편차: {best_case['표준편차']:.2f}°C (값이 낮을수록 균일)")
    print(f"  - 전체 평균 온도: {best_case['평균']:.2f}°C (목표 {target_temp}°C)")
    print(f"  - 목표 온도와의 차이: {best_case['목표온도와의 차이']:.2f}°C")

def analyze_case_details(sensor_df: pd.DataFrame, case_name: str, time_desc: str):
    """
    (신규) 특정 케이스에 대해 3가지 기준의 상세 Boxplot 비교 및 통계량을 출력합니다.
    """
    print("\n" + "="*80 + f"\n[{case_name} - {time_desc}] 상세 분석\n" + "="*80)
    
    case_df = sensor_df[(sensor_df['Case'] == case_name) & (sensor_df['TimeDesc'] == time_desc)].copy()
    if case_df.empty:
        print(f"해당 조건의 데이터가 없습니다.")
        return

    # --- 데이터 가공 ---
    # 1. 선단/중단/후단 구분
    case_df['위치 (선/중/후)'] = pd.cut(case_df['Sensor'], bins=[0, 3, 6, 9], labels=['선단 (S1~3)', '중단 (S4~6)', '후단 (S7~9)'])
    
    # 2. 상부/하부 소재 구분
    case_df['소재 위치 (상/하)'] = np.where(case_df['Bloom'] <= 6, '하부소재 (B1~6)', '상부소재 (B7~12)')
    
    # 3. 센서 높이 구분
    def categorize_sensor_height(sensor_num):
        if sensor_num in [1, 4, 7]: return '하부 (S1,4,7)'
        if sensor_num in [2, 5, 8]: return '중단 (S2,5,8)'
        if sensor_num in [3, 6, 9]: return '상부 (S3,6,9)'
        return None
    case_df['센서 위치 (상/중/하)'] = case_df['Sensor'].apply(categorize_sensor_height)

    # --- 시각화 ---
    fig, axes = plt.subplots(1, 3, figsize=(24, 8))
    fig.suptitle(f'[{case_name} - {time_desc}] 상세 온도 분포 비교', fontsize=20, y=1.02)

    # 비교표 1
    sns.boxplot(ax=axes[0], x='위치 (선/중/후)', y='Value', data=case_df, palette='Reds')
    axes[0].set_title('비교표 1: 선단/중단/후단별 온도 분포', fontsize=14)
    axes[0].set_xlabel('')
    axes[0].set_ylabel('온도 (°C)')
    axes[0].grid(True, alpha=0.6)

    # 비교표 2
    sns.boxplot(ax=axes[1], x='소재 위치 (상/하)', y='Value', data=case_df, palette='Blues')
    axes[1].set_title('비교표 2: 상부/하부 소재별 온도 분포', fontsize=14)
    axes[1].set_xlabel('')
    axes[1].set_ylabel('')
    axes[1].grid(True, alpha=0.6)

    # 비교표 3
    sns.boxplot(ax=axes[2], x='센서 위치 (상/중/하)', y='Value', data=case_df, palette='Greens')
    axes[2].set_title('비교표 3: 센서 높이별 온도 분포', fontsize=14)
    axes[2].set_xlabel('')
    axes[2].set_ylabel('')
    axes[2].grid(True, alpha=0.6)

    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()

    # --- 통계량 출력 ---
    stats1 = case_df.groupby('위치 (선/중/후)')['Value'].describe()
    stats2 = case_df.groupby('소재 위치 (상/하)')['Value'].describe()
    stats3 = case_df.groupby('센서 위치 (상/중/하)')['Value'].describe()

    print("\n[비교표 1: 선단/중단/후단별 통계]\n")
    print(stats1.to_string(float_format='{:.2f}'.format))
    print("\n" + "-"*50)
    
    print("\n[비교표 2: 상부/하부 소재별 통계]\n")
    print(stats2.to_string(float_format='{:.2f}'.format))
    print("\n" + "-"*50)

    print("\n[비교표 3: 센서 높이별 통계]\n")
    print(stats3.to_string(float_format='{:.2f}'.format))


# ==============================================================================
# Step 5. [메인 실행 로직]
# ==============================================================================
def main():
    """
    데이터 분석 및 시각화 전체 프로세스를 실행합니다.
    """
    # --- 1. 분석 환경 설정 ---
    excel_file_path = "250718_가열로 시뮬레이션 results4_가공.xlsx"
    target_temperature = 1280.0
    if not os.path.exists(excel_file_path):
        print(f"오류: 엑셀 파일을 찾을 수 없습니다 - {excel_file_path}")
        return

    case_meta_data = {
        'Case': [
            'Case 1', 'Case 2', 'Case 2*', 'Case 3', 'Case 4', 'Case 5', 'Case 6', 'Case 7', 'Case 8', 'Case 8 냉각',
            'Case 9', 'Case 10', 'Case 11', 'Case 11 추가', 'Case 12', 'Case 13', 'Case 13 추가', 'Case 14', 'Case 14 추가',
            'Case 15', 'Case 15 냉각', 'Case 16', 'Case 17', 'Case 18', 'Case 18 냉각', 'Case 19', 'Case 20'
        ],
        'Sheet_Name': [
            'Case 1', 'Case 2', 'Case 2"', 'Case 3', 'Case 4', 'Case 5', 'Case 6', 'Case 7', 'Case 8', 'Case 8 냉각',
            'Case 9', 'Case 10', 'Case 11', 'Case 11 추가', 'Case 12', 'Case 13', 'Case 13 추가', 'Case 14', 'Case 14 추가',
            'Case 15', 'Case 15 냉각', 'Case 16', 'Case 17', 'Case 18', 'Case 18 냉각', 'Case 19', 'Case 20'
        ],
        'Furnace': [
            'Batch-Type', 'Batch-Type', 'Batch-Type', 'Batch-Type', 'Batch-Type', 'Batch-Type', 'Batch-Type', 'Batch-Type', 'Batch-Type', 'Batch-Type',
            'Batch-Type', 'Batch-Type', 'Batch-Type', 'Batch-Type', 'Batch-Type', 'Batch-Type', 'Batch-Type', 'Batch-Type', 'Batch-Type',
            'Soaking-Pit', 'Soaking-Pit', 'Soaking-Pit', 'Soaking-Pit', 'Soaking-Pit', 'Soaking-Pit', 'Soaking-Pit', 'Soaking-Pit'
        ],
        '작업방향': [
            'Longitudinal', 'Longitudinal', 'Longitudinal', 'Longitudinal', 'Longitudinal', 'Longitudinal', 'Longitudinal', 'Longitudinal', 'Longitudinal', 'Longitudinal',
            'Longitudinal', 'Longitudinal', 'Transverse', 'Transverse', 'Transverse', 'Transverse', 'Transverse', 'Transverse', 'Transverse',
            '수직 장입', '수직 장입', '수직 장입', '수직 장입', '수직 장입', '수직 장입', '수직 장입', '수직 장입'
        ],
        '적치단수': [ '2단', '2단', '2단', '1단', '2단', '2단', '2단', '2단', '1단', '1단', '2단', '2단', '2단', '2단', '1단', '2단', '2단', '2단', '2단', 'X', 'X', 'X', 'X', 'X', 'X', 'X', 'X' ],
        '서포트높이': [ '250mm', '250mm', '250mm', '250mm', '250mm', '250mm', '470mm', '470mm', '470mm', '470mm', '470mm', '470mm', '250mm', '250mm', '250mm', '250mm', '250mm', '470mm', '470mm', 'X', 'X', 'X', 'X', 'X', 'X', 'X', 'X' ],
        '작업방법': [ '2단(2-1)', '2단(2-5hr-1)', '2단(2-5hr-1)', '1단', '2단(2-1)', '2단(2-5hr-1)', '2단(2-1)', '2단(2-5hr-1)', '1단', '1단', '2단(2-1)', '2단(2-5hr-1)', '2단(2-1)', '2단(2-1)', '1단', '2단(2-5hr-1)', '2단(2-1)', '2단(2-1)', '2단(2-1)', '-', '-', '-', '-', '-', '-', '-', '-' ],
        '제어': [ '1Zone', '1Zone', '1Zone', '1Zone', '2Zone', '2Zone', '1Zone', '1Zone', '1Zone', '1Zone', '2Zone', '2Zone', '1Zone', '1Zone', '1Zone', '2Zone', '2Zone', '2Zone', '2Zone', '1Zone', '1Zone', '1Zone', '2Zone', '2Zone', '2Zone', '1Zone', '2Zone' ],
        '냉각': [ '-', '-', '7zone 37332초 후 온도고정', '-', '-', '-', '-', '-', '-', '대기 냉각 추가 (600초)', '-', '-', '추가 해석', '추가 해석', '-', '추가 해석', '추가 해석', '추가 해석', '추가 해석', '-', '대기 냉각 추가 (600초)', '-', '625', '-', '대기 냉각 추가 (600초)=625', '-', '-' ]
    }
    case_meta_df = pd.DataFrame(case_meta_data)
    
    # --- 1a. [신규] 분석 대상 케이스 선택 ---
    selected_cases = [
        'Case 1', 'Case 2', 'Case 2*', 'Case 3', 'Case 4', 'Case 5', 'Case 6', 'Case 7', 'Case 8', 'Case 8 냉각',
        'Case 9', 'Case 10', 'Case 11', 'Case 11 추가', 'Case 12', 'Case 13', 'Case 13 추가', 'Case 14', 'Case 14 추가',
        'Case 15', 'Case 15 냉각', 'Case 16', 'Case 17', 'Case 18', 'Case 18 냉각', 'Case 19', 'Case 20'
        # 'Case 1', 'Case 4', 'Case 5', 'Case 6', 'Case 9', 'Case 10', 
        # 'Case 11 추가', 'Case 13 추가', 'Case 14 추가', 
        # 'Case 15', 'Case 16', 'Case 17', 'Case 18', 'Case 19', 'Case 20',
        # 'Case 8 냉각', 'Case 15 냉각', 'Case 18 냉각'
    ]
    case_meta_df = case_meta_df[case_meta_df['Case'].isin(selected_cases)].copy()
    case_meta_df.reset_index(drop=True, inplace=True)

    case_sheets = case_meta_df['Sheet_Name'].tolist()

    time_points = { "4hr 시점": 47860.0, "가열완료 시점": 57382.5, "냉각완료 시점": 57382.5 + 600 }
    
    # --- 2. 엑셀 시트 미리 읽어오기 ---
    print("엑셀 파일 로딩 시작...")
    all_sheets_data = {}
    for sheet in case_sheets:
        try:
            df = pd.read_excel(excel_file_path, sheet_name=sheet, engine='openpyxl', header=find_header_row(excel_file_path, sheet))
            all_sheets_data[sheet] = df
        except Exception as e:
            print(f"'{sheet}' 시트 로딩 실패: {e}")

    print(f"엑셀 파일 로딩 완료. ({len(all_sheets_data)}/{len(case_sheets)} 시트)")
    if not all_sheets_data: return

    # --- 3. 데이터 추출 및 병합 ---
    print("데이터 추출 및 가공 시작...")
    bloom_sensor_all = []
    cooling_curve_all = []
    long_heating_cases = case_meta_df[case_meta_df['작업방법'].str.contains('2-5hr', na=False)]['Sheet_Name'].tolist()
    special_completion_time = 75382.5

    for sheet, df in all_sheets_data.items():
        meta_info = case_meta_df[case_meta_df['Sheet_Name'] == sheet].iloc[0]
        heat_time = special_completion_time if sheet in long_heating_cases else time_points["가열완료 시점"]
        
        for time_desc, time_val in [("4hr 시점", time_points["4hr 시점"]), ("가열완료 시점", heat_time)]:
            result = extract_bloom_sensor_table(df, sheet, time_val)
            if result is not None:
                result["TimeDesc"] = time_desc
                bloom_sensor_all.append(result)

        is_cooling_case = '냉각' in meta_info['Case']
        if is_cooling_case:
            curve_data = extract_cooling_curve_data(df, meta_info['Case'], 0, 600)
            if curve_data is not None:
                cooling_curve_all.append(curve_data)

    if not bloom_sensor_all:
        print("추출된 데이터가 없습니다. 분석을 중단합니다.")
        return
    
    bloom_sensor_df = pd.concat(bloom_sensor_all, ignore_index=True)
    bloom_sensor_df.rename(columns={'Case': 'Sheet_Name'}, inplace=True)
    bloom_sensor_df = pd.merge(bloom_sensor_df, case_meta_df, on="Sheet_Name")

    # --- 4. 요약 통계 생성 및 데이터 분리 ---
    summary_cols = ['Case', 'TimeDesc', 'Bloom', 'Furnace', '작업방향', '서포트높이', '제어', '적치단수', '작업방법']
    bloom_summary = (
        bloom_sensor_df.groupby([col for col in summary_cols if col in bloom_sensor_df.columns])
        .agg(평균=('Value', 'mean'), 내부ΔT=('Value', lambda x: np.nanmax(x) - np.nanmin(x) if x.notna().any() else np.nan))
        .reset_index()
    )
    
    cooling_case_names = case_meta_df[case_meta_df['Case'].str.contains('냉각')]['Case'].tolist()
    heating_summary = bloom_summary[~bloom_summary['Case'].isin(cooling_case_names)]
    heating_sensor_df = bloom_sensor_df[~bloom_sensor_df['Case'].isin(cooling_case_names)]

    # --- 5. 히트맵 색상 범위 계산 ---
    vmin_color = bloom_summary['평균'].quantile(0.05) if bloom_summary['평균'].notna().any() else 1000
    vmax_color = bloom_summary['평균'].quantile(0.95) if bloom_summary['평균'].notna().any() else 1300

    # --- 6. 분석 실행 및 시각화 ---
    case_order = sorted(case_meta_df['Case'].unique(), key=natural_sort_key)
    analysis_time_points = ["4hr 시점", "가열완료 시점"]
    for time_point in analysis_time_points:
        print("\n" + "="*50 + f"\n[{time_point} 케이스 분석]\n" + "="*50)
        current_heating_summary = heating_summary[heating_summary['TimeDesc'] == time_point]
        current_sensor_df = heating_sensor_df[heating_sensor_df['TimeDesc'] == time_point]
        
        if not current_heating_summary.empty:
            avg_min_tp = current_heating_summary['평균'].min() - 10
            avg_max_tp = current_heating_summary['평균'].max() + 10
            delta_max_tp = current_heating_summary['내부ΔT'].max() + 5 if current_heating_summary['내부ΔT'].notna().any() else 50
            avg_range_tp = (avg_min_tp, avg_max_tp)
            delta_range_tp = (0, delta_max_tp)
            
            sensor_min_tp = current_sensor_df['Value'].min() - 20
            sensor_max_tp = current_sensor_df['Value'].max() + 20
            sensor_range_tp = (sensor_min_tp, sensor_max_tp)

            plot_grouped_comparison_heatmaps(current_heating_summary, case_meta_df, '작업방향', time_point, vmin_color, vmax_color)
            plot_advanced_boxplots(current_heating_summary, time_point, avg_range_tp, delta_range_tp, case_order)
            plot_case_deviation_boxplots(current_sensor_df, time_point, target_temperature, sensor_range_tp, case_order)
        else:
            print(f"'{time_point}'에 대한 분석 데이터가 없습니다.")

    if cooling_curve_all:
        print("\n" + "="*50 + "\n[냉각 완료 케이스 분석]\n" + "="*50)
        cooling_curve_df = pd.concat(cooling_curve_all, ignore_index=True)
        plot_cooling_curves(cooling_curve_df)

    # --- 7. 온도 균일도 분석 및 최적 케이스 선정 ---
    if not heating_sensor_df.empty:
        analyze_temperature_uniformity(heating_sensor_df, heating_summary, target_temperature)

    # --- 8. [신규] 선택된 케이스에 대한 상세 분석 실행 ---
    # 수정: 모든 선택된 가열 케이스에 대해 4hr, 가열완료 시점 모두 분석
    detailed_analysis_cases = [case for case in selected_cases if '냉각' not in case]
    for case in detailed_analysis_cases:
        for time_point in analysis_time_points:
             analyze_case_details(bloom_sensor_df, case, time_point)


if __name__ == '__main__':
    main()
```
